### Step 1: Import Libraries and Load Data


- Import the `pandas`, `matplotlib.pyplot`, and `seaborn` libraries.


- You will begin with loading the dataset. You can use the pyfetch method if working on JupyterLite. Otherwise, you can use pandas' read_csv() function directly on their local machines or cloud environments.


```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Stack Overflow survey dataset
data_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv'
df = pd.read_csv(data_url)

# Display the first few rows of the dataset
df.head()
```

### Step 2: Examine the Structure of the Data


- Display the column names, data types, and summary information to understand the data structure.

- Objective: Gain insights into the dataset's shape and available variables.


```python
# 1. Display column names
print("Column Names:")
print(df.columns)

# 2. Display data types of each column
print("\nData Types:")
print(df.dtypes)

# 3. Display summary information
print("\nDataFrame Information:")
df.info()

# 4. Display dataset shape (rows, columns)
print("\nDataset Shape:")
print(df.shape)
```

### Step 3: Handle Missing Data


- Identify missing values in the dataset.

- Impute or remove missing values as necessary to ensure data completeness.



```python
# Count missing values in each column
missing_values = df.isnull().sum()

# Display columns with missing values only
missing_values = missing_values[missing_values > 0]

print(missing_values.sort_values(ascending=False))
```

```python
missing_percent = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
'Missing Count': df.isnull().sum(),
'Missing %': missing_percent
})

print(
missing_summary[missing_summary['Missing Count'] > 0]
.sort_values('Missing %', ascending=False)
)
```

```python
# IMPUTE NUMERICAL COLUMNS
num_cols = df.select_dtypes(include=['int64', 'float64']).columns

for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
#IMPUTE CATEGORICAL COLUMNS
cat_cols = df.select_dtypes(include=['object']).columns

for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
print("Remaining missing values:")
print(df.isnull().sum().sum())

df.info()
```

### Step 4: Analyze Key Columns


- Examine key columns such as `Employment`, `JobSat` (Job Satisfaction), and `YearsCodePro` (Professional Coding Experience).

- **Instruction**: Calculate the value counts for each column to understand the distribution of responses.



```python
# Employment distribution
print("Employment Distribution:")
print(df['Employment'].value_counts(dropna=False))

# Job Satisfaction distribution
print("\nJob Satisfaction Distribution:")
print(df['JobSat'].value_counts(dropna=False))

# Professional Coding Experience distribution
print("\nYearsCodePro Distribution:")
print(df['YearsCodePro'].value_counts(dropna=False))
```

### Step 5: Visualize Job Satisfaction (Focus on JobSat)


- Create a pie chart or KDE plot to visualize the distribution of `JobSat`.

- Provide an interpretation of the plot, highlighting key trends in job satisfaction.


```python
# Ensure JobSat is numeric
jobsat = pd.to_numeric(df['JobSat'], errors='coerce')

plt.figure(figsize=(10, 6))
sns.kdeplot(jobsat.dropna(), fill=True)

plt.title('Distribution of Job Satisfaction (JobSat)')
plt.xlabel('Job Satisfaction Score')
plt.ylabel('Density')
plt.grid(alpha=0.3)
plt.show()
```

### Step 6: Programming Languages Analysis


- Compare the frequency of programming languages in `LanguageHaveWorkedWith` and `LanguageWantToWorkWith`.
  
- Visualize the overlap or differences using a Venn diagram or a grouped bar chart.


```python
#COMPARE LANGUANGE FREQUENCIES
import pandas as pd
from collections import Counter

# Languages respondents have worked with
worked_langs = Counter()

for langs in df['LanguageHaveWorkedWith'].dropna():
    worked_langs.update([lang.strip() for lang in langs.split(';')])

worked_df = pd.DataFrame(
    worked_langs.items(),
    columns=['Language', 'WorkedWithCount']
)

# Languages respondents want to work with
want_langs = Counter()

for langs in df['LanguageWantToWorkWith'].dropna():
    want_langs.update([lang.strip() for lang in langs.split(';')])

want_df = pd.DataFrame(
    want_langs.items(),
    columns=['Language', 'WantToWorkWithCount']
)

# Merge counts
lang_compare = pd.merge(
    worked_df,
    want_df,
    on='Language',
    how='outer'
).fillna(0)

lang_compare = lang_compare.sort_values(
    'WorkedWithCount',
    ascending=False
)

print(lang_compare.head(20))
```

```python
# Top 15 languages by worked-with count
top_langs = lang_compare.nlargest(15, 'WorkedWithCount')

plot_df = top_langs.melt(
id_vars='Language',
value_vars=['WorkedWithCount', 'WantToWorkWithCount'],
var_name='Category',
value_name='Count'
)

plt.figure(figsize=(14, 7))

sns.barplot(
data=plot_df,
x='Language',
y='Count',
hue='Category'
)

plt.title('Programming Languages: Worked With vs Want To Work With')
plt.xlabel('Language')
plt.ylabel('Number of Respondents')
plt.xticks(rotation=45)
plt.legend(title='')

plt.tight_layout()
plt.show()
```

### Step 7: Analyze Remote Work Trends


- Visualize the distribution of RemoteWork by region using a grouped bar chart or heatmap.


```python
# Top 15 countries
top_countries = df['Country'].value_counts().head(15).index

heat_df = df[df['Country'].isin(top_countries)]

# Cross-tabulation
remote_country = pd.crosstab(
heat_df['Country'],
heat_df['RemoteWork']
)

plt.figure(figsize=(12, 8))

sns.heatmap(
remote_country,
annot=True,
fmt='d',
cmap='Blues'
)

plt.title('Remote Work Preferences by Country')
plt.xlabel('Remote Work Type')
plt.ylabel('Country')

plt.tight_layout()
plt.show()
```

### Step 8: Correlation between Job Satisfaction and Experience


- Analyze the correlation between overall job satisfaction (`JobSat`) and `YearsCodePro`.
  
- Calculate the Pearson or Spearman correlation coefficient.


```python
corr_df = df[['YearsCodePro', 'JobSat']].copy()

corr_df['YearsCodePro'] = corr_df['YearsCodePro'].replace({
'Less than 1 year': 0.5,
'More than 50 years': 50
})

corr_df['YearsCodePro'] = pd.to_numeric(
corr_df['YearsCodePro'],
errors='coerce'
)

corr_df = corr_df.dropna()

# Pearson
pearson_corr = corr_df['YearsCodePro'].corr(corr_df['JobSat'])

# Spearman (without scipy)
spearman_corr = (
corr_df['YearsCodePro'].rank()
.corr(corr_df['JobSat'].rank())
)

print(f"Pearson Correlation : {pearson_corr:.4f}")
print(f"Spearman Correlation: {spearman_corr:.4f}")
```

### Step 9: Cross-tabulation Analysis (Employment vs. Education Level)


- Analyze the relationship between employment status (`Employment`) and education level (`EdLevel`).

- **Instruction**: Create a cross-tabulation using `pd.crosstab()` and visualize it with a stacked bar plot if possible.


```python
# Top education levels and employment types
top_ed = df['EdLevel'].value_counts().head(10).index
top_emp = df['Employment'].value_counts().head(10).index

filtered_df = df[
(df['EdLevel'].isin(top_ed)) &
(df['Employment'].isin(top_emp))
]

# Cross-tabulation
edu_emp = pd.crosstab(
filtered_df['EdLevel'],
filtered_df['Employment']
)

edu_emp_pct = pd.crosstab(
filtered_df['EdLevel'],
filtered_df['Employment'],
normalize='index'
) * 100

print(edu_emp_pct.round(2))
```

```python
edu_emp_pct.plot(
kind='bar',
stacked=True,
figsize=(14, 8),
colormap='tab20'
)

plt.title('Employment Status by Education Level')
plt.xlabel('Education Level')
plt.ylabel('Percentage of Respondents')
plt.legend(
title='Employment',
bbox_to_anchor=(1.05, 1),
loc='upper left'
)

plt.tight_layout()
plt.show()
```
